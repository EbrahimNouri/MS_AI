۲ (نظارت شده - Classification):

داده‌ها دارای برچسب هستند (مثبت/منفی)

می‌توانیم Accuracy یا F1-Score را محاسبه کنیم چون نظارت شده است و پاسخ صحیح را می‌دانیم

 ۶ (غیرنظارتی - Clustering):

داده‌ها هیچ برچسبی ندارند

خوشه‌بندی الگوهای پنهان را کشف می‌کند، نه اینکه کلاس‌های از پیش تعیین‌شده را پیش‌بینی کند

نمی‌توان Accuracy یا F1-Score استفاده کرد چون پاسخ درست مشخص نیست

معیارهای مناسب: Inertia، Silhouette Score، Davies-Bouldin Index



# B

مثال عددی:

۹۹۰ تراکنش سالم (کلاس ۰)

۱۰ تراکنش مشکوک (کلاس ۱)

مدل ساختگی که همیشه «سالم» پیش‌بینی می‌کند:

Accuracy = ۹۹۰/۱۰۰۰ = ۹۹٪ (خیلی بالا!)

اما Recall = ۰/۱۰ = ۰ (هیچ تراکنش مشکوکی را پیدا نکرده!)

نتیجه: Accuracy بالا گمراه‌کننده است. باید به Recall (درصد تشخیص درست تراکنش‌های مشکوک) توجه کرد که در این مدل صفر است.

# G

 تفاوت خروجی Regression و Classification
ویژگی	Regression	Classification
خروجی	عدد پیوسته	دسته/کلاس
مثال	۳۵.۶ دقیقه	«مثبت» یا «منفی»
معیارها	MAE, MSE, R²	Accuracy, Precision, Recall, F1
دقت معیارها	فاصله خطا را اندازه می‌گیرند	درصد یا نرخ تشخیص را اندازه می‌گیرند
تأثیر بر معیار ارزیابی:

در Regression: خطای عددی اهمیت دارد (مثلاً خطای ۵ دقیقه با خطای ۲۰ دقیقه فرق دارد)

در Classification: فقط درست/غلط بودن اهمیت دارد، میزان خطا معنا ندارد
