```python
import numpy as np
from sklearn.neighbors import KNeighborsClassifier

np.random.seed(42)

# =========================
# Generate Random Data
# =========================

n = 100

# Class 0 -> Normal distribution
X_class0 = np.random.normal(
    loc=2,
    scale=1,
    size=(n, 2)
)

# Class 1 -> Normal distribution
X_class1 = np.random.normal(
    loc=6,
    scale=1,
    size=(n, 2)
)

X = np.vstack((X_class0, X_class1))
y = np.array([0] * n + [1] * n)

# Point that we want to classify
x = np.array([4, 4])


# ==========================================================
# PART 1
# Bayes Rule - Distribution is known
# ==========================================================

def normal_pdf(x, mean, std):

    return (
        1 / (std * np.sqrt(2 * np.pi))
        * np.exp(
            -0.5 * ((x - mean) / std) ** 2
        )
    )


# Mean and standard deviation of each class

mean_0 = np.mean(X_class0, axis=0)
std_0 = np.std(X_class0, axis=0)

mean_1 = np.mean(X_class1, axis=0)
std_1 = np.std(X_class1, axis=0)


# P(x | class)

p_x_given_0 = np.prod(
    normal_pdf(x, mean_0, std_0)
)

p_x_given_1 = np.prod(
    normal_pdf(x, mean_1, std_1)
)


# P(class)

p_0 = 0.5
p_1 = 0.5


# Bayes Rule

p_x = (
    p_x_given_0 * p_0
    + p_x_given_1 * p_1
)

p_0_given_x = (
    p_x_given_0 * p_0
) / p_x

p_1_given_x = (
    p_x_given_1 * p_1
) / p_x


print("===== Bayes =====")

print("P(class 0 | x):", p_0_given_x)
print("P(class 1 | x):", p_1_given_x)


# ==========================================================
# PART 2
# KNN - Distribution is unknown
# ==========================================================

k = 5

knn = KNeighborsClassifier(
    n_neighbors=k
)

knn.fit(X, y)

# P(class | x)
probabilities = knn.predict_proba(
    x.reshape(1, -1)
)

print("\n===== KNN =====")

print("Classes:", knn.classes_)
print("P(class | x):", probabilities[0])

print("Predicted class:",
      knn.predict(x.reshape(1, -1))[0])
```
