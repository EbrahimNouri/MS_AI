```python
import numpy as np
from pprint import pprint
```

```python
bib_numbers = np.array([101, 102, 103, 104, 105, 106])
times_5k = np.array([22.3, 25.1, 21.8, 26.4, 23.0, 24.7])
```

# 1

```python
bib_numbers_2 = np.array([107, 108])
times_5k_2 = np.array([20.5, 27.9])

bib_numbers = np.hstack((bib_numbers, bib_numbers_2))
times_5k = np.hstack((times_5k, times_5k_2))

backup_bib_numbers = bib_numbers.copy()
backup_times_5k = times_5k.copy()

print(bib_numbers)
print(times_5k)
```

# 2

```python
np.random.seed(0)
runners = np.column_stack((bib_numbers, times_5k))
np.random.shuffle(runners)
bib_numbers = runners[:, 0]
times_5k = runners[:, 1]
pprint(bib_numbers)
pprint(times_5k)
pprint(backup_bib_numbers)
pprint(backup_times_5k)
```

# 3

```python
# print(runners)

sorted_indices = np.argsort(runners[:, 1])

# print(f"sorted_indices {sorted_indices}")

print(f"rank{'':<3} bib{'':<5} time{'':>6}")
for rank, idx in enumerate(sorted_indices, 1):
  bib = bib_numbers[idx]
  time = times_5k[idx]
  print(f"  {rank:<3}  {bib:<5}  {time:>6.1f}")
```

# 4

```python
bib = 104
bib_index = np.where(bib_numbers == bib)[0][0]
rank = np.where(sorted_indices == bib_index)[0][0] + 1
time = times_5k[bib_index]
print(f"rank: {rank}  bib: {int(bib)}  time: {time:.1f}")
```
