```python
import numpy as np
```

```python
# rng = np.random.default_rng(seed=42)

np.random.seed(0)
weather_data = np.random.randn(2, 8, 4) * 5 + 20
weather_data
```

```python
def analyze_day(data: np.ndarray) -> bool:
  return data.shape == (4, 8)
```

# 1

```python
day2 = weather_data[1]
print(f"day2.shape: {day2.shape}")
day2_t = day2.T
print(f"day2_t.shape: {day2_t.shape}")
print(f"analyze_day: {analyze_day(day2_t)}")
```
