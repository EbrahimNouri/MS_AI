```python
import numpy as np
```

```python
def scale_range_data(data):
  min_vals = np.min(data, axis=0)
  max_vals = np.max(data, axis=0)
  return (data - min_vals) / (max_vals - min_vals)
```

# 1

```python
# zamane pokht(min), celery, tondi(1_10), tedade mavade avaliye
recipes = np.array([
  [15, 350, 2, 5],
  [45, 600, 7, 10],
  [10, 200, 0, 3],
  [30, 450, 5, 7],
  [60, 800, 8, 12]
])
recipe_names = ["Salad", "Curry", "Toast", "Pasta", "Stew"]

recipes_scaled = scale_range_data(recipes)
print(recipes_scaled)

```

# 2

```python
users = np.array([
  [10, 250, 1, 4],
  [50, 700, 8, 11],
  [25, 400, 4, 6]
])

users_scaled = scale_range_data(users)
print(users_scaled)
```

# 3

```python
distances = np.sqrt(np.sum((users_scaled[:, np.newaxis, :] - recipes_scaled[np.newaxis, :, :]) ** 2, axis=2))

print(f"distances: \n{distances}")
```

# 4

```python
for i in range(len(users)):
  sorted_indices = np.argsort(distances[i])
  sorted_recipes = np.take(recipe_names, sorted_indices)
  sorted_distances = distances[i][sorted_indices]

  print(f"\nuser {i + 1}:")
  print(f"  rank {sorted_recipes[0]}: {sorted_recipes[0]} (dist = {sorted_distances[0]:.4f})")
```

# 5

```python
for i in range(len(users)):
  sorted_indices = np.argsort(distances[i])
  sorted_recipes = np.take(recipe_names, sorted_indices)
  sorted_distances = distances[i][sorted_indices]

  print(f"\nuser {i + 1}:")
  for rank, (recipe, dist) in enumerate(zip(sorted_recipes, sorted_distances), 1):
    print(f"  rank {rank}: {recipe} (dist = {dist:.4f})")
```
