```python
import numpy as np
```

```python
importance = np.array([3, 8, 1, 9, 4, 7])
# print(importance.shape)

importance_reshaped1 = importance.reshape(-1, 1)
print(f"importance_reshaped1.shape: {importance_reshaped1.shape}")
print(f"importance_reshaped1: \n{importance_reshaped1}")

importance_reshaped2 = np.expand_dims(importance, axis=1)
print(f"importance_reshaped2.shape: {importance_reshaped2.shape}")
print(f"importance_reshaped2: \n{importance_reshaped2}")

importance_reshaped3 = importance[:, np.newaxis]
print(f"importance_reshaped3.shape: {importance_reshaped3.shape}")
print(f"importance_reshaped3: \n{importance_reshaped3}")

```

```python
notes = np.array([5, 10, 15, 20, 25])
pinned = notes[1:3]
pinned[0] = 999
print(f"without copy: {notes}")

notes = np.array([5, 10, 15, 20, 25])
pinned = notes[1:3].copy()
pinned[0] = 999
print(f"with copy {notes}")
```
