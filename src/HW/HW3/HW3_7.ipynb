```python
import numpy as np
```

```python
X_messages = np.array([
  [12, 0, 1],
  [45, 5, 8],
  [8, 0, 0],
  [30, 3, 4]
])
w = np.array([0.1, 0.8, 0.5])
b = -2.0
```

# 1

```python
y = (X_messages @ w) + b
print(f"y = " f"{y}")
print(f"w = {w}")
print(f"b = {b}")
```

# 2

```python
y_ReLu = np.maximum(y, 0)
print(f"y = " f"{y_ReLu}")
```

# 3

```python
threshold = 5
predictions = np.where(y_ReLu >= threshold, "high energy", "calm")

print("-"*30)
for i, msg in enumerate(X_messages):
    print(f"message  {i+1}: {msg}")
    print(f"  raw score: {y[i]:.2f}")
    print(f"  after ReLu: {y_ReLu[i]:.2f}")
    print(f"  prediction: {predictions[i]}")
    print("-"*30)
```
