```python
import numpy as np
```

```python
#quiz, mian, final
scores = np.array([
  [18, 15, 20],
  [12, 14, 16],
  [20, 19, 18],
  [10, 8, 15]
])
scheme_A = np.array([0.5, 0.3, 0.2])
scheme_B = np.array([0.2, 0.3, 0.5])
scheme_C = np.array([0.1, 0.2, 0.7])
```

# 1

```python
weight_matrix = np.array([scheme_A, scheme_B, scheme_C])
print(f"weight_matrix: \n{weight_matrix}")
```

# 2

```python
final_scores = scores @ weight_matrix.T
print(f"final_scores: \n{final_scores}")
print(f"final_scores shape: \n{final_scores.shape}")
```

# 3

```python
arg_max = np.argmax(final_scores, axis=1)
labels = ['A', 'B', 'C']
for i, j in enumerate(arg_max):
  print(f"{i}: {labels[j]}")
```

# 4

```python
means = np.mean(final_scores, axis=0)
print(f"means:\n{means}")
max_mean = np.max(means)
max_arg = np.argmax(means)
print(f"max_arg: {max_arg}\nmax_mean: {max_mean}")
```

# 5

```python
# w1 + w2 + w3 = 1
# w3 = 2 * w1

quiz_weight = 0.25
final_weight = 2 * quiz_weight
midterm_weight = 1 - (quiz_weight + final_weight)

scheme_D = np.array((quiz_weight, midterm_weight, final_weight))

print(f"scheme_D: \n{scheme_D}")
weight_matrix = np.vstack((weight_matrix, scheme_D))
print(f"weight_matrix: \n{weight_matrix}")
```
