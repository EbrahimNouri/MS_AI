## کد اولیه
سلول زیر را اجرا کنید.

</div>

```python
import numpy as np
import matplotlib.pyplot as plt

rng = np.random.default_rng(42)

n_samples = 240

X = rng.normal(size=(n_samples, 3))

true_w = np.array([
    1.6,
    -1.1,
    0.8
])

true_b = -0.2

true_logits = X @ true_w + true_b

true_prob = 1 / (1 + np.exp(-true_logits))

y = (
    rng.random(n_samples)
    < true_prob
).astype(float)

feature_names = [
    "waiting_time",
    "first_response_quality",
    "reopen_history"
]

print("X shape:", X.shape)
print("y shape:", y.shape)
print("Positive rate:", y.mean())
```

## توابع آماده‌ی شرکت

توابع زیر توسط تیم قبلی آماده شده‌اند و **نباید تغییر داده شوند**.

در این فعالیت قرار نیست فرمول داخلی Gradient را استخراج کنید. تمرکز روی **تفسیر Gradient و استفاده از آن در Gradient Descent** است.

</div>

```python
def sigmoid(z):
    z = np.clip(z, -50, 50)
    return 1 / (1 + np.exp(-z))


def dataset_bce(y, p):
    eps = 1e-12
    p = np.clip(p, eps, 1 - eps)

    loss = -np.mean(
        y * np.log(p)
        +
        (1 - y) * np.log(1 - p)
    )

    return loss


def company_training_service(X, y, w, b):
    z = X @ w + b
    p = sigmoid(z)

    loss = dataset_bce(y, p)

    error_signal = p - y

    grad_w = (
        X.T @ error_signal
    ) / len(X)

    grad_b = np.mean(error_signal)

    return {
        "probability": p,
        "loss": loss,
        "error_signal": error_signal,
        "grad_w": grad_w,
        "grad_b": grad_b
    }
```

# بخش نهایی ـ آیا واقعاً فهمیدید مدل چگونه یاد می‌گیرد؟

بدون استفاده از تعریف‌های حفظی پاسخ دهید:

1. مدل در ابتدای Training چه چیزی را هنوز نمی‌داند؟
2. Forward Pass چه چیزی تولید می‌کند؟
3. Loss چه چیزی درباره‌ی Prediction به ما می‌گوید؟
4. چرا هدف Training کاهش Loss است؟
5. Gradient چه اطلاعاتی برای تغییر پارامترها فراهم می‌کند؟
6. Learning Rate چه چیزی را کنترل می‌کند؟
7. چرا Learning Rate بسیار کوچک می‌تواند مشکل‌ساز باشد؟
8. چرا Learning Rate بسیار بزرگ می‌تواند Training را ناپایدار کند؟
9. Gradient Descent چگونه از Gradient و Learning Rate برای تغییر پارامترها استفاده می‌کند؟
10. چرا Mini-Batch در بسیاری از پروژه‌های Deep Learning یک انتخاب عملی است؟
