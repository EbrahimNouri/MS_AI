```python
# بخش 2

data = np.array([[32,55,9,7],
                [28,35,8,5],
                [18,50,6,4],
                [30,20,5,3],
                [22,75,7,6],
                [25,40,9,5],
                [20,45,8,4],
                [17,45,4,3]])
# 2B
mean = np.mean(data, axis=0)
print(mean)
max_m = np.max(data, axis=0)
print(max_m)
min_m = np.min(data, axis=0)
print(min_m)
std = np.std(data, axis=0)
print(std)
var = np.var(data,axis=0)
print(var)

scaled_data = ((data - min_m) / (max_m - min_m))
print(scaled_data)
# اسکیلینگ باعث میشود که فاصله ها در یک رنج (0 -1) قرار بگیرند و مقایسه فاصله ها به درستی انجام شود.

# 2c
ideal = np.array([26, 40, 8, 4])
scaled_ideal = ((ideal - min_m) / (max_m - min_m))

euclidean = np.linalg.norm(scaled_data - scaled_ideal,axis= 1)
print(f"euclidean: {euclidean} ")
print(f" min euclidean:{np.min(euclidean)}")
min_index = np.argmin(euclidean)
print(min_index)
print(data[min_index])
print('---' * 30)
max_index = np.argmax(euclidean)
print(max_index)
print(data[max_index])


# cosine_similarity = ((np.dot(scaled_data,scaled_ideal)) / (np.linalg.norm(scaled_data,axis=1) * np.linalg.norm(scaled_ideal)))
norms_data = np.linalg.norm(scaled_data, axis=1) # Shape: (8,)
norm_ideal = np.linalg.norm(scaled_ideal) # Shape: ()

cosine_similarity = np.dot(scaled_data, scaled_ideal) / (norms_data * norm_ideal)
print(f"cosine_similarity: {cosine_similarity}")
print(f"max cosine: {np.argmax(cosine_similarity)}")
print(f"euclidean: {euclidean} ")
print(f"min euclidean: {np.argmin(euclidean)} ")

print("--" * 30)

# 2D
weights = np.array([0.1, 0.2, 0.3, 0.4])

scaled_data_weighted = scaled_data * weights
# print(f"scaled_data_weighted:\n{scaled_data_weighted}")

scaled_ideal_weighted = scaled_ideal * weights
# print(f'scaled_ideal before weighted:\n{scaled_ideal}')
# print(f'scaled_ideal_weighted:\n{scaled_ideal_weighted}')

cosine_similarity_weighted = ((np.dot(scaled_data_weighted,scaled_ideal_weighted)) / (np.linalg.norm(scaled_data_weighted,axis=1) * np.linalg.norm(scaled_ideal_weighted)))
print(f"cosine_similarity_weighted:\n {cosine_similarity_weighted}")
```
